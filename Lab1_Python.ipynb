Task 1

```python
from random import randint
n = 31
mas1 = tuple(randint(10, 35) for i in range(n))
print("Температура в течение августа:")
for i in mas1:
    print(i, end=" ")
print()
max_temp = mas1[0]
for i in mas1:
    if i > max_temp:
        max_temp = i
print("Максимальная температура августа:", max_temp)
```

Task 2

```python
from random import randint
n, m, k = 3, 4, 5
A = tuple(tuple(randint(1, 9) for j in range(m)) for i in range(n))
B = tuple(tuple(randint(1, 9) for j in range(k)) for i in range(m))
print("Матрица A: ")
for i in A:
    print(*i)
print("\nМатрица B: ")
for i in B:
    print(*i)
C = tuple(
    tuple(
        sum(A[i][t] * B[t][j] for t in range(m))
        for j in range(k)
    )
    for i in range(n)
)
print("\nРезультат (матрица C = A * B):")
for i in C:
    print(*i)
```

Task 3

```python
from random import randint
m = 15
list1 = [randint(1, 5) for j in range(m)]
print("Список чисел: ")
for i in list1:
    print(i, end=" ")
print()
count = 0
for i in range(len(list1)):
    for j in range(i + 1, len(list1)):
        if list1[i] == list1[j]:
            count += 1
print("Количество пар равных элементов: ", count)
```

Task 4

```python
from random import randint
list1 = [randint(10, 100) for _ in range(randint(10, 21))]
list2 = [randint(10, 100) for _ in range(randint(10, 21))]
set1 = set(list1)
set2 = set(list2)
print("Первый список: ")
for i in list1:
    print(i, end=" ")
print()
print("Второй список: ")
for i in list2:
    print(i, end=" ")
print()
common = set1 & set2
print("Числа, входящие в оба списка, в порядке возрастания: ")
for i in sorted(common):
    print(i, end=" ")
print()
```

Task 5

```python
import re
text = input("Введите строку: ")
words = re.findall(r'[A-Za-zА-Яа-яЁё]+', text.lower())
word_count = dict()
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1
min_count = min(word_count.values())
rare_words = [word for word, count in word_count.items() if count == min_count]
result = min(rare_words)
print("Слово, которое встречается реже всего: ", result)
```
